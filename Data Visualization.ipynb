```python
x = np.linspace(0,5,11)
x
```

```python
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Sqaure of x")
plt.xlabel("x axis")
plt.ylabel("y axis sqaure of x")
```

```python
# subplots in matpoltlib 
```

```python
plt.subplot(2,2,1)
plt.plot(x,x)
plt.subplot(2,2,2) 
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(y,y**2)
```

### Matrix plot

**Heat Map**

```python
flights = sns.load_dataset('flights')
flights
```

```python
tips =  sns.load_dataset('tips')
tips
```

```python
tipscorr = tips[['total_bill','tip','size']]
```

```python
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(),annot=True)
```

**Cluster Map**

```python
sns.clustermap(tipscorr.corr())
```

**Pivot Table Heat Map**

```python
pytFlights = flights.pivot_table(values = 'passengers', index = 'month', columns = 'year')
```

```python
pytFlights
```

```python
sns.heatmap(pytFlights)
```
