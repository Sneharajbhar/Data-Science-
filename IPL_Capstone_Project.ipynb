```python
#1. Who scored the highest number of runs in the dataset?
df[df['Runs'] == df['Runs'].max()]
```

```python
#2. Which player has the highest batting average (AVG)?
df[df['AVG'] == df['AVG'].max()]['Player Name']
```

```python
#3. Who has the highest strike rate (SR) among all players?
df[df['SR'] == df['SR'].max()]['Player Name']
```

```python
#4. How many players have scored at least one century (100s > 0)?
df[df['100s'] > 0]['Player Name'].count()
```

```python
#5. Which player has hit the most sixes (6s)?
df[df['6s'] == df['6s'].max()]['Player Name']
```

```python
#6. Among players who scored more than 600 runs, who has the best strike rate?
df[df['Runs'] > 600].sort_values('SR', ascending=False).iloc[0]['Player Name']
```

```python
#7. Which player has the most half-centuries (50s) without scoring a century?
df[df['100s'] == 0].loc[df['50s'].idxmax(), 'Player Name']
```

```python
df2.isnull().sum()
```

```python
# which team won the most matches

match_wins = df2['match_winner'].value_counts() 
sns.barplot(y = match_wins.index,x = match_wins.values,palette= 'viridis')
plt.title("Most match win by team")
```

```python
# Toss decision trends 

sns.countplot(x = df2['toss_decision'],palette='rainbow')
plt.title("Toss decision trends")
```

```python
# Toss winner vs match winner

count = df2[df2['toss_winner'] == df2['match_winner']]['match_id'].count()

percentage = (count*100)/df2.shape[0]
percentage.round(2)
```

```python
sns.countplot(df2['won_by'])
```

### Key player performance

```python
# 1 most "player of the match" Awards

count = df2['player_of_the_match'].value_counts().head(10)
sns.barplot(x = count.values,y = count.index,palette='mako')
plt.title("Top 10 players with man of the match")
```

```python
# 2 top scorers
high = df2.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind = 'barh')
```

```python
df2['highest_wickets'] = df2['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df2['highest_wickets'] = df2['highest_wickets'].astype(int)
df2.head()
```

```python
# top bowlers
top_bowlers = df2.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers
```

```python
top_bowlers.plot(kind = 'barh')
```
